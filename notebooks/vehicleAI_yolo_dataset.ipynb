```python
# ============================================================
# CELL 16 — Single-frame YOLO detection test
# ============================================================

cap = cv2.VideoCapture(
    str(VIDEO_PATH)
)

success, frame = cap.read()

cap.release()

if not success:
    raise RuntimeError(
        "Could not read first frame."
    )

results = model.predict(
    source=frame,
    classes=VEHICLE_CLASSES,
    conf=0.25,
    imgsz=640,
    device=0,
    verbose=False
)

result = results[0]

print(
    "Detections:",
    len(result.boxes)
)

for box in result.boxes:

    class_id = int(
        box.cls.item()
    )

    confidence = float(
        box.conf.item()
    )

    coords = (
        box.xyxy[0]
        .detach()
        .cpu()
        .numpy()
    )

    print(
        VEHICLE_CLASS_IDS.get(
            class_id,
            str(class_id)
        ),
        f"conf={confidence:.3f}",
        "bbox=",
        coords.astype(int)
    )
```

```python
# ============================================================
# CELL 17 — YOLO + ByteTrack smoke test
# ============================================================

import time

SMOKE_TEST_FRAMES = 300

cap = cv2.VideoCapture(
    str(VIDEO_PATH)
)

if not cap.isOpened():
    raise RuntimeError(
        "Could not open video."
    )

track_records = []

frame_number = 0

start_time = time.time()

while frame_number < SMOKE_TEST_FRAMES:

    success, frame = cap.read()

    if not success:
        break

    frame_number += 1

    results = model.track(
        source=frame,
        persist=True,
        tracker="bytetrack.yaml",
        classes=VEHICLE_CLASSES,
        conf=0.25,
        imgsz=640,
        device=0,
        verbose=False
    )

    result = results[0]

    if result.boxes is None:
        continue

    if result.boxes.id is None:
        continue

    track_ids = (
        result.boxes.id
        .detach()
        .cpu()
        .numpy()
        .astype(int)
    )

    boxes = (
        result.boxes.xyxy
        .detach()
        .cpu()
        .numpy()
    )

    class_ids = (
        result.boxes.cls
        .detach()
        .cpu()
        .numpy()
        .astype(int)
    )

    confidences = (
        result.boxes.conf
        .detach()
        .cpu()
        .numpy()
    )

    for track_id, box, class_id, confidence in zip(
        track_ids,
        boxes,
        class_ids,
        confidences
    ):

        x1, y1, x2, y2 = box

        track_records.append({
            "frame_number": frame_number,
            "track_id": int(track_id),
            "class_id": int(class_id),
            "class_name": VEHICLE_CLASS_IDS[
                int(class_id)
            ],
            "confidence": float(confidence),
            "x1": float(x1),
            "y1": float(y1),
            "x2": float(x2),
            "y2": float(y2)
        })

cap.release()

elapsed = time.time() - start_time

tracks_df = pd.DataFrame(
    track_records
)

print(
    "Frames processed:",
    frame_number
)

print(
    "Track detections:",
    len(tracks_df)
)

print(
    "Unique track IDs:",
    tracks_df["track_id"].nunique()
    if len(tracks_df)
    else 0
)

print(
    "Runtime:",
    round(elapsed, 2),
    "seconds"
)

if len(tracks_df):
    print(
        "\nVehicle classes:"
    )
    print(
        tracks_df[
            "class_name"
        ].value_counts()
    )
```

```python
# ============================================================
# CELL 19 — Full YOLO + ByteTrack + vehicle crop extraction
# ============================================================

import cv2
import json
import time
import numpy as np
import pandas as pd
from pathlib import Path
from tqdm.auto import tqdm

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

EXPERIMENT_ROOT = (
    PROCESSED_ROOT
    / "experiments"
    / "S01_c001"
)

VIDEO_PATH = (
    EXPERIMENT_ROOT
    / "vdo.avi"
)

TRACKS_CSV = (
    EXPERIMENT_ROOT
    / "tracks.csv"
)

TRACKS_JSON = (
    EXPERIMENT_ROOT
    / "tracks.json"
)

CROPS_ROOT = (
    EXPERIMENT_ROOT
    / "crops"
)

CROPS_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# Video information
# ------------------------------------------------------------

cap = cv2.VideoCapture(
    str(VIDEO_PATH)
)

if not cap.isOpened():
    raise RuntimeError(
        "Could not open video."
    )

fps = cap.get(
    cv2.CAP_PROP_FPS
)

total_frames = int(
    cap.get(
        cv2.CAP_PROP_FRAME_COUNT
    )
)

width = int(
    cap.get(
        cv2.CAP_PROP_FRAME_WIDTH
    )
)

height = int(
    cap.get(
        cv2.CAP_PROP_FRAME_HEIGHT
    )
)

print("Video:", VIDEO_PATH)
print("Resolution:", width, "x", height)
print("FPS:", fps)
print("Frames:", total_frames)

# ------------------------------------------------------------
# Tracking
# ------------------------------------------------------------

track_records = []

# Keep a small number of crops per track.
# We don't want to save every frame.
MAX_CROPS_PER_TRACK = 8

saved_crop_counts = {}

frame_number = 0

start_time = time.time()

with tqdm(
    total=total_frames,
    desc="YOLO + ByteTrack"
) as pbar:

    while True:

        success, frame = cap.read()

        if not success:
            break

        frame_number += 1

        results = model.track(
            source=frame,
            persist=True,
            tracker="bytetrack.yaml",
            classes=VEHICLE_CLASSES,
            conf=0.25,
            imgsz=640,
            device=0,
            verbose=False
        )

        result = results[0]

        if (
            result.boxes is None
            or result.boxes.id is None
        ):
            pbar.update(1)
            continue

        track_ids = (
            result.boxes.id
            .detach()
            .cpu()
            .numpy()
            .astype(int)
        )

        boxes = (
            result.boxes.xyxy
            .detach()
            .cpu()
            .numpy()
        )

        class_ids = (
            result.boxes.cls
            .detach()
            .cpu()
            .numpy()
            .astype(int)
        )

        confidences = (
            result.boxes.conf
            .detach()
            .cpu()
            .numpy()
        )

        for (
            track_id,
            box,
            class_id,
            confidence
        ) in zip(
            track_ids,
            boxes,
            class_ids,
            confidences
        ):

            track_id = int(track_id)
            class_id = int(class_id)

            x1, y1, x2, y2 = (
                box.astype(int)
            )

            # Clamp coordinates
            x1 = max(0, min(x1, width - 1))
            y1 = max(0, min(y1, height - 1))
            x2 = max(0, min(x2, width))
            y2 = max(0, min(y2, height))

            # Ignore invalid boxes
            if x2 <= x1 or y2 <= y1:
                continue

            track_records.append({
                "frame_number": frame_number,
                "track_id": track_id,
                "class_id": class_id,
                "class_name": VEHICLE_CLASS_IDS[
                    class_id
                ],
                "confidence": float(confidence),
                "x1": x1,
                "y1": y1,
                "x2": x2,
                "y2": y2
            })

            # ------------------------------------------------
            # Save representative crops
            # ------------------------------------------------

            current_count = saved_crop_counts.get(
                track_id,
                0
            )

            if current_count < MAX_CROPS_PER_TRACK:

                crop = frame[
                    y1:y2,
                    x1:x2
                ]

                if crop.size > 0:

                    track_dir = (
                        CROPS_ROOT
                        / f"track_{track_id:05d}"
                    )

                    track_dir.mkdir(
                        parents=True,
                        exist_ok=True
                    )

                    crop_path = (
                        track_dir
                        / f"frame_{frame_number:06d}.jpg"
                    )

                    cv2.imwrite(
                        str(crop_path),
                        crop
                    )

                    saved_crop_counts[
                        track_id
                    ] = current_count + 1

        pbar.update(1)

cap.release()

elapsed = time.time() - start_time

# ------------------------------------------------------------
# Convert to DataFrame
# ------------------------------------------------------------

tracks_df = pd.DataFrame(
    track_records
)

# ------------------------------------------------------------
# Save tracking table
# ------------------------------------------------------------

tracks_df.to_csv(
    TRACKS_CSV,
    index=False
)

print("\n==========================================")
print("FULL S01_c001 TRACKING COMPLETE")
print("==========================================")

print(
    "Frames processed:",
    frame_number
)

print(
    "Tracking detections:",
    len(tracks_df)
)

print(
    "Unique tracks:",
    tracks_df["track_id"].nunique()
    if not tracks_df.empty
    else 0
)

print(
    "Runtime:",
    round(elapsed / 60, 2),
    "minutes"
)

print(
    "CSV:",
    TRACKS_CSV
)

print(
    "Crop directory:",
    CROPS_ROOT
)
```

```python
# ============================================================
# CELL 20 — Build VehicleIQ track metadata
# ============================================================

track_metadata = {}

for track_id, group in (
    tracks_df.groupby("track_id")
):

    group = group.sort_values(
        "frame_number"
    )

    track_metadata[
        str(track_id)
    ] = {
        "track_id": int(track_id),

        "first_frame": int(
            group["frame_number"].min()
        ),

        "last_frame": int(
            group["frame_number"].max()
        ),

        "frame_count": int(
            group["frame_number"].nunique()
        ),

        "class_distribution": (
            group["class_name"]
            .value_counts()
            .to_dict()
        ),

        "mean_detection_confidence": float(
            group["confidence"].mean()
        ),

        "crops": [
            str(p.relative_to(EXPERIMENT_ROOT))
            for p in sorted(
                (
                    CROPS_ROOT
                    / f"track_{int(track_id):05d}"
                ).glob("*.jpg")
            )
        ]
    }

with open(
    TRACKS_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        track_metadata,
        f,
        indent=2
    )

print(
    "Track JSON saved:"
)

print(
    TRACKS_JSON
)

print(
    "\nTotal tracks:",
    len(track_metadata)
)

print("\nFirst track:")

first_track_id = next(
    iter(track_metadata)
)

print(
    json.dumps(
        track_metadata[first_track_id],
        indent=2
    )
)
```

```python
# ============================================================
# CELL 21 — Calculate image-space vehicle velocity
# ============================================================

tracks_df = pd.read_csv(
    TRACKS_CSV
)

# ------------------------------------------------------------
# Bounding-box center
# ------------------------------------------------------------

tracks_df["center_x"] = (
    tracks_df["x1"] +
    tracks_df["x2"]
) / 2.0

tracks_df["center_y"] = (
    tracks_df["y1"] +
    tracks_df["y2"]
) / 2.0

# ------------------------------------------------------------
# Sort by track and frame
# ------------------------------------------------------------

tracks_df = tracks_df.sort_values(
    [
        "track_id",
        "frame_number"
    ]
).reset_index(
    drop=True
)

# ------------------------------------------------------------
# Frame-to-frame displacement
# ------------------------------------------------------------

tracks_df["dx_pixels"] = (
    tracks_df
    .groupby("track_id")["center_x"]
    .diff()
)

tracks_df["dy_pixels"] = (
    tracks_df
    .groupby("track_id")["center_y"]
    .diff()
)

# ------------------------------------------------------------
# Time between frames
# ------------------------------------------------------------

tracks_df["dt_seconds"] = (
    tracks_df
    .groupby("track_id")["frame_number"]
    .diff()
    / fps
)

# ------------------------------------------------------------
# Pixel velocity components
# ------------------------------------------------------------

tracks_df["vx_pixels_per_sec"] = (
    tracks_df["dx_pixels"] /
    tracks_df["dt_seconds"]
)

tracks_df["vy_pixels_per_sec"] = (
    tracks_df["dy_pixels"] /
    tracks_df["dt_seconds"]
)

# ------------------------------------------------------------
# Magnitude
# ------------------------------------------------------------

tracks_df["speed_pixels_per_sec"] = np.sqrt(
    tracks_df[
        "vx_pixels_per_sec"
    ] ** 2
    +
    tracks_df[
        "vy_pixels_per_sec"
    ] ** 2
)

# First observation of each track has no velocity
tracks_df[
    [
        "dx_pixels",
        "dy_pixels",
        "dt_seconds",
        "vx_pixels_per_sec",
        "vy_pixels_per_sec",
        "speed_pixels_per_sec"
    ]
] = tracks_df[
    [
        "dx_pixels",
        "dy_pixels",
        "dt_seconds",
        "vx_pixels_per_sec",
        "vy_pixels_per_sec",
        "speed_pixels_per_sec"
    ]
].replace(
    [np.inf, -np.inf],
    np.nan
)

# ------------------------------------------------------------
# Save updated tracking table
# ------------------------------------------------------------

tracks_df.to_csv(
    TRACKS_CSV,
    index=False
)

print(
    "Updated tracking table saved."
)

print(
    "\nColumns added:"
)

for column in [
    "center_x",
    "center_y",
    "dx_pixels",
    "dy_pixels",
    "dt_seconds",
    "vx_pixels_per_sec",
    "vy_pixels_per_sec",
    "speed_pixels_per_sec"
]:

    print(
        " -",
        column
    )
```

```python
# ============================================================
# CELL 23 — Temporal representative-frame sampling
# ============================================================

import cv2
import json
from pathlib import Path

MAX_REPRESENTATIVE_CROPS = 8

TEMPORAL_CROPS_ROOT = (
    EXPERIMENT_ROOT
    / "temporal_crops"
)

TEMPORAL_CROPS_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# Open video
# ------------------------------------------------------------

cap = cv2.VideoCapture(
    str(VIDEO_PATH)
)

if not cap.isOpened():
    raise RuntimeError(
        "Could not open video."
    )

temporal_crop_metadata = {}

for track_id, group in (
    tracks_df.groupby("track_id")
):

    group = group.sort_values(
        "frame_number"
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # Select frame positions across the entire track
    # --------------------------------------------------------

    count = len(group)

    sample_count = min(
        MAX_REPRESENTATIVE_CROPS,
        count
    )

    sample_positions = np.linspace(
        0,
        count - 1,
        sample_count
    ).round().astype(int)

    selected_rows = group.iloc[
        sample_positions
    ]

    track_dir = (
        TEMPORAL_CROPS_ROOT
        / f"track_{int(track_id):05d}"
    )

    track_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    selected_crops = []

    # --------------------------------------------------------
    # Extract selected frames
    # --------------------------------------------------------

    for _, row in selected_rows.iterrows():

        frame_number = int(
            row["frame_number"]
        )

        cap.set(
            cv2.CAP_PROP_POS_FRAMES,
            frame_number - 1
        )

        success, frame = cap.read()

        if not success:
            continue

        x1 = int(row["x1"])
        y1 = int(row["y1"])
        x2 = int(row["x2"])
        y2 = int(row["y2"])

        crop = frame[
            y1:y2,
            x1:x2
        ]

        if crop.size == 0:
            continue

        crop_path = (
            track_dir
            / f"frame_{frame_number:06d}.jpg"
        )

        cv2.imwrite(
            str(crop_path),
            crop
        )

        selected_crops.append({
            "frame_number": frame_number,
            "crop_path": str(
                crop_path.relative_to(
                    EXPERIMENT_ROOT
                )
            )
        })

    temporal_crop_metadata[
        str(track_id)
    ] = selected_crops

cap.release()

print(
    "Tracks processed:",
    len(temporal_crop_metadata)
)

print(
    "Temporal crops stored in:",
    TEMPORAL_CROPS_ROOT
)
```

```python
# ============================================================
# CELL 26 — Load SigLIP 2
# ============================================================

from transformers import (
    AutoProcessor,
    AutoModel
)

MODEL_ID = (
    "google/siglip2-base-patch16-224"
)

device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

if device.type != "cuda":
    raise RuntimeError(
        "GPU is required for this step."
    )

processor = AutoProcessor.from_pretrained(
    MODEL_ID
)

model = AutoModel.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16
).to(device)

model.eval()

print(
    "Model:",
    MODEL_ID
)

print(
    "GPU:",
    torch.cuda.get_device_name(0)
)
```

```python
# ============================================================
# CELL 27 — SigLIP image encoder
# ============================================================

import torch.nn.functional as F
from PIL import Image

def encode_images(
    images,
    batch_size=32
):

    if not isinstance(images, list):
        images = [images]

    all_embeddings = []

    for start in range(
        0,
        len(images),
        batch_size
    ):

        batch_images = images[
            start:start + batch_size
        ]

        inputs = processor(
            images=batch_images,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
            if torch.is_tensor(value)
        }

        with torch.inference_mode():

            output = model.get_image_features(
                **inputs
            )

        if hasattr(
            output,
            "pooler_output"
        ):
            embeddings = output.pooler_output
        else:
            embeddings = output

        embeddings = F.normalize(
            embeddings,
            p=2,
            dim=-1
        )

        all_embeddings.append(
            embeddings
        )

    return torch.cat(
        all_embeddings,
        dim=0
    )
```

```python
# ============================================================
# CELL 28 — Vehicle embeddings
# ============================================================

import numpy as np
from tqdm.auto import tqdm

vehicle_embedding_list = []
vehicle_embedding_ids = []

for track_id in tqdm(
    sorted(
        temporal_crop_metadata,
        key=lambda x: int(x)
    ),
    desc="Encoding vehicle tracks"
):

    crop_records = (
        temporal_crop_metadata[
            track_id
        ]
    )

    images = []

    valid_paths = []

    for item in crop_records:

        crop_path = (
            EXPERIMENT_ROOT
            / item["crop_path"]
        )

        if not crop_path.exists():
            continue

        image = Image.open(
            crop_path
        ).convert("RGB")

        images.append(image)
        valid_paths.append(
            crop_path
        )

    if not images:
        continue

    frame_embeddings = encode_images(
        images
    )

    # Mean pooling across temporal views
    vehicle_embedding = (
        frame_embeddings
        .float()
        .mean(dim=0)
    )

    vehicle_embedding = F.normalize(
        vehicle_embedding,
        p=2,
        dim=0
    )

    vehicle_embedding_list.append(
        vehicle_embedding
        .cpu()
        .numpy()
    )

    vehicle_embedding_ids.append(
        int(track_id)
    )

vehicle_embeddings = np.asarray(
    vehicle_embedding_list,
    dtype=np.float32
)

print(
    "Vehicle embeddings:",
    vehicle_embeddings.shape
)

print(
    "Vehicles encoded:",
    len(vehicle_embedding_ids)
)
```

```python
# ============================================================
# CELL 31 — SigLIP text encoder
# ============================================================

def encode_text(
    texts,
    batch_size=32
):

    if isinstance(texts, str):
        texts = [texts]

    all_embeddings = []

    for start in range(
        0,
        len(texts),
        batch_size
    ):

        batch_texts = texts[
            start:start + batch_size
        ]

        inputs = processor(
            text=batch_texts,
            padding="max_length",
            truncation=True,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
            if torch.is_tensor(value)
        }

        with torch.inference_mode():

            output = model.get_text_features(
                **inputs
            )

        if hasattr(
            output,
            "pooler_output"
        ):
            embeddings = output.pooler_output
        else:
            embeddings = output

        embeddings = F.normalize(
            embeddings,
            p=2,
            dim=-1
        )

        all_embeddings.append(
            embeddings
        )

    return torch.cat(
        all_embeddings,
        dim=0
    )
```

```python
# ============================================================
# CELL 32 — Zero-shot vehicle attributes
# ============================================================

ATTRIBUTE_PROMPTS = {

    "color": [
        "a red vehicle",
        "a blue vehicle",
        "a white vehicle",
        "a black vehicle",
        "a gray vehicle",
        "a silver vehicle",
        "a green vehicle",
        "a yellow vehicle",
        "a brown vehicle"
    ],

    "body_style": [
        "a sedan",
        "an SUV",
        "a hatchback",
        "a pickup truck",
        "a van",
        "a bus",
        "a motorcycle",
        "a truck"
    ],

    "orientation": [
        "a vehicle viewed from the front",
        "a vehicle viewed from the rear",
        "a vehicle viewed from the left side",
        "a vehicle viewed from the right side",
        "a vehicle viewed from the front-left",
        "a vehicle viewed from the front-right",
        "a vehicle viewed from the rear-left",
        "a vehicle viewed from the rear-right"
    ]
}

# Encode all prompts once
attribute_prompt_embeddings = {}

for attribute, prompts in (
    ATTRIBUTE_PROMPTS.items()
):

    attribute_prompt_embeddings[
        attribute
    ] = encode_text(prompts)

print(
    "Zero-shot attribute prompts encoded."
)

for attribute, embeddings in (
    attribute_prompt_embeddings.items()
):

    print(
        attribute,
        embeddings.shape
    )
```

```python
import json
import torch
import torch.nn.functional as F
import numpy as np
from pathlib import Path

# ============================================================
# CELL 33 — Build final VehicleIQ JSON records
# ============================================================

vehicle_records = {}

embedding_by_track = {
    str(track_id): vehicle_embeddings[i]
    for i, track_id in enumerate(
        vehicle_embedding_ids
    )
}

for track_id, track_info in (
    track_metadata.items()
):

    if track_id not in embedding_by_track:
        continue

    vehicle_vector = torch.tensor(
        embedding_by_track[track_id],
        dtype=torch.float16, # Changed from torch.float32 to torch.float16
        device=device
    ).unsqueeze(0)

    # --------------------------------------------------------
    # Zero-shot attributes
    # --------------------------------------------------------

    predicted_attributes = {}

    for attribute, prompt_embeddings in (
        attribute_prompt_embeddings.items()
    ):

        similarities = (
            vehicle_vector
            @ prompt_embeddings.T
        ).squeeze(0)

        best_index = int(
            torch.argmax(
                similarities
            ).item()
        )

        prompts = ATTRIBUTE_PROMPTS[
            attribute
        ]

        predicted_attributes[
            attribute
        ] = {
            "value":
                prompts[best_index],
            "confidence":
                float(
                    similarities[
                        best_index
                    ].item()
                )
        }

    # --------------------------------------------------------
    # Tracker class
    # --------------------------------------------------------

    class_distribution = (
        track_info[
            "class_distribution"
        ]
    )

    vehicle_type = max(
        class_distribution,
        key=class_distribution.get
    )

    # --------------------------------------------------------
    # Final record
    # --------------------------------------------------------

    vehicle_records[
        track_id
    ] = {

        "vehicle_id":
            f"S01_c001_v_{int(track_id):05d}",

        "source": {
            "video_id": "S01_c001",
            "video_path": str(
                VIDEO_PATH
            )
        },

        "tracking": {
            "track_id":
                track_info["track_id"],

            "first_frame":
                track_info["first_frame"],

            "last_frame":
                track_info["last_frame"],

            "frame_count":
                track_info["frame_count"]
        },

        "vehicle": {

            "type": {
                "value":
                    vehicle_type,

                "confidence":
                    track_info[
                        "mean_detection_confidence"
                    ]
            },

            "color":
                predicted_attributes[
                    "color"
                ],

            "body_style":
                predicted_attributes[
                    "body_style"
                ],

            "orientation":
                predicted_attributes[
                    "orientation"
                ]
        },

        "velocity":
            track_info[
                "velocity"
            ],

        "visual": {

            "representative_crops":
                track_info["crops"]
        },

        "embeddings": {

            "model":
                MODEL_ID,

            "path":
                str(
                    VIDEO_EMBEDDINGS_PATH
                )
        }
    }

FINAL_VEHICLE_JSON = (
    EXPERIMENT_ROOT
    / "vehicle_records.json"
)

with open(
    FINAL_VEHICLE_JSON,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        vehicle_records,
        f,
        indent=2
    )

print(
    "Final VehicleIQ records:",
    len(vehicle_records)
)

print(
    "Saved:",
    FINAL_VEHICLE_JSON
)
```

```python
# ============================================================
# CELL 38 — End-to-end semantic search
# ============================================================

def semantic_search_video(
    query,
    top_k=10
):

    text_embedding = encode_text(
        query
    )

    query_vector = (
        text_embedding
        .detach()
        .cpu()
        .numpy()
        .astype(np.float32)
    )

    scores, indices = video_index.search(
        query_vector,
        min(
            top_k,
            video_index.ntotal
        )
    )

    results = []

    for rank, (
        score,
        index_position
    ) in enumerate(
        zip(
            scores[0],
            indices[0]
        ),
        start=1
    ):

        track_id = (
            vehicle_embedding_ids[
                index_position
            ]
        )

        vehicle_id = (
            f"S01_c001_v_{track_id:05d}"
        )

        row = cursor.execute("""
        SELECT json_record
        FROM vehicles
        WHERE vehicle_id = ?
        """, (
            vehicle_id,
        )).fetchone()

        if row is None:
            continue

        record = json.loads(
            row[0]
        )

        results.append({
            "rank": rank,
            "similarity": float(score),
            "vehicle_id": vehicle_id,
            "track_id": track_id,
            "record": record
        })

    return results
```
